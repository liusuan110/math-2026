```python
import sys, os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 让 notebook 能 import 仓库里的复用模块（data_prep / plotting 等）
CODE = os.path.abspath(os.path.join(os.getcwd(), '..'))
for p in (os.path.join(CODE, 'preprocessing'), os.path.join(CODE, 'common')):
    if p not in sys.path:
        sys.path.insert(0, p)
import data_prep as dp
from plotting import setup_style, save_fig
setup_style()  # 中文字体 + 论文级风格
print('环境就绪')
```

```python
# 1) 读数据（按题改路径；xlsx 用 read_excel，csv 用 read_csv）
DATA_PATH = '../../data/附件1.xlsx'   # TODO: 改成实际附件
df = pd.read_excel(DATA_PATH) if DATA_PATH.endswith(('.xlsx', '.xls')) else pd.read_csv(DATA_PATH)
print('形状:', df.shape)
df.head()
```

```python
# 2) 缺失值报告
dp.missing_report(df)
```

```python
# 3) 描述统计 + 类型
display(df.describe(include='all').T)
df.dtypes
```

```python
# 4) 填补缺失 + 异常值检测
clean = dp.fill_missing(df, strategy='median')
outliers = dp.detect_outliers(clean, method='iqr')
print('各列异常值个数:')
print(outliers.sum())
```

```python
# 5) 相关性热图（存到 figures/）
corr = dp.correlation(clean, save_heatmap='../../figures/探索_相关性热图.png')
corr.round(2)
```

```python
# 6) 各数值列分布直方图
num = clean.select_dtypes(include='number')
axes = num.hist(figsize=(12, 8), bins=30)
plt.tight_layout()
plt.show()
```
